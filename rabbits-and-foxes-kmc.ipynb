Same farm, same rate constants: 400 rabbits and 200 foxes at the start, and

$$\frac{dR}{dt} = k_1 R - k_2 R F \qquad\qquad \frac{dF}{dt} = k_3 R F - k_4 F$$

with $k_1 = 0.015$ day$^{-1}$, $k_2 = 0.00004$ day$^{-1}$ foxes$^{-1}$,
$k_3 = 0.0004$ day$^{-1}$ rabbits$^{-1}$ and $k_4 = 0.04$ day$^{-1}$.

In the `solve_ivp` homework you found that the foxes fall to **under two** in each
trough. The ODEs don't mind: 1.6 foxes carry on and breed. Real foxes can't be
fractional, and two foxes can easily both die before they meet a rabbit. So now we
treat every rabbit and fox as an individual, and every birth and death as a random
event. That is **kinetic Monte Carlo** (KMC).

### The events

There are four events. Each has a rate, in events per day, which changes as the
populations change:

| Event | Rate | Effect |
|---|---|---|
| a rabbit is born | $k_1 R$ | $R \to R + 1$ |
| a rabbit is eaten | $k_2 R F$ | $R \to R - 1$ |
| a fox is born | $k_3 R F$ | $F \to F + 1$ |
| a fox dies | $k_4 F$ | $F \to F - 1$ |

The rates are easy to calculate, so we use the **rejection-free** algorithm, which you
worked out in class from
[Wikipedia's Kinetic Monte Carlo page](https://en.wikipedia.org/wiki/Kinetic_Monte_Carlo).
Writing it down in your own words is the first part of task 1.

````python
import time

import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

k1 = 0.015  # 1/day
k2 = 0.00004  # 1/day/fox
k3 = 0.0004  # 1/day/rabbit
k4 = 0.04  # 1/day

R0, F0 = 400, 200  # rabbits and foxes at t = 0
t_end = 600  # days
````

The deterministic answer, for comparison. This is the `solve_ivp` homework's
solution, at a tight tolerance.

````python
def dydt(t, y):
    R, F = y
    return [k1 * R - k2 * R * F, k3 * R * F - k4 * F]

ode = solve_ivp(dydt, (0, t_end), [R0, F0], rtol=1e-10, atol=1e-8, dense_output=True)
t_ode = np.linspace(0, t_end, 3001)
R_ode, F_ode = ode.sol(t_ode)

plt.plot(t_ode, R_ode, label="rabbits (ODE)")
plt.plot(t_ode, F_ode, label="foxes (ODE)")
plt.xlabel("time (days)")
plt.ylabel("population")
plt.legend()
plt.show()
````

## 1. One trajectory

**(a) The algorithm, in your own words.** Before you write any code, write the
rejection-free algorithm as numbered steps, for rabbits and foxes: what each of
Wikipedia's symbols ($k$, $r_{ki}$, $R_{ki}$, $Q_k$, $u$, $u'$) becomes here, how the
next event is chosen, and how far the clock moves. Then work the **first step by
hand**, from 400 rabbits and 200 foxes, with $u = 0.5$ and $u' = 0.3$: which event
happens, and how many minutes pass?

*Answer:*

````python
1. k is the current state of the system, i.e the numbers of rabbits and foxes right now, (R,F)
2.r_ki is the rate of event i in that state. Since we have four events, i = 1 to 4. The events are: rabbits reproduce, foxes eat rabbits, foxes reproduce, and foxes die. The rates of these events are given by the equations:Rabbit reproduce: r_k1 = k1 * R, rabbits eaten by foxes: r_k2 = k2 * R * F, foxes reproduce: r_k3 = k3 * R * F, foxes die: r_k4 = k4 * F. The total rate of all events is the sum of the individual rates: r_total = r_k1 + r_k2 + r_k3 + r_k4
3.R_ki is the cumulative rate: R_k1 = r_k1, R_k2 = r_k1 + r_k2, and so on.
4.Q_k is the total rate of all events, which is the last cumulative rate, R_k4.
5. is a random number in (0, 1] that picks which event happens.
6. u′ is a second random number in (0, 1] that picks how long until it happens.
The steps:
1.t=0
k = (R0, F0) = (400, 200)  # Initial state with R0 rabbits and F0 foxes
2.Calculate the four rates and their cumulative sums
3. Get Q, the total rate. If Q = 0, stop.
4. Draw u. Pick the event i where R_k,i−1 < uQ ≤ R_ki. Imagine laying the four rates end to end on a ruler of length Q and seeing where uQ lands. Faster events take up more of the ruler, so they get picked more often.
5.Apply that event: add or remove one rabbit or fox.
6.Draw u′ and move the clock forward by Δt = ln(1/u′)/Q. A higher total rate means shorter waits.
7. If t < t_end, go back to step 2.

At R=400 and F=200, k1 = 0.015  # 1/day
k2 = 0.00004  # 1/day/fox
k3 = 0.0004  # 1/day/rabbit
k4 = 0.04  # 1/day the rates are:
- Rabbit reproduce: r_k1 = k1 * R = 0.015 * 400 = 6.0
- Rabbits eaten by foxes: r_k2 = k2 * R * F = 0.00004 * 400 * 200 = 3.2
- Foxes reproduce: r_k3 = k3 * R * F = 0.0004 * 400 * 200 = 32.0
- Foxes die: r_k4 = k4 * F = 0.04 * 200 = 8.0     
cumulative rates, R_ki = 
- R_k1 = r_k1 = 6.0
- R_k2 = r_k1 + r_k2 = 6.0 + 3.2 = 9.2
- R_k3 = r_k1 + r_k2 + r_k3 = 9.2 + 32.0 = 41.2
- R_k4 = r_k1 + r_k2 + r_k3 + r_k4 = 41.2 + 8.0 = 49.2
Q_k = R_k4 = 49.2 events/day
uQ = 0.5 * 49.2 = 24.6, That falls between 9.2 and 41.2, so a fox is born, and F goes from 200 to 201.
Δt = ln(1/0.3)/49.2 = 0.0203 days 
0.0203 days is about 29.2 minutes, so the new state is (R, F) = (400, 201) at t = 0.0203 days.
````

**(b) The function.** Write

```python
def kmc(R0, F0, t_end, rng):
    ...
    return times, rabbits, foxes   # three NumPy arrays, one entry per event
```

that runs your algorithm from $t = 0$ until the clock passes `t_end`. If the total
rate is ever zero, nothing more can happen, so stop there.

Run it once. How many events were there, and how long did it take? Plot rabbits and
foxes against time, on top of the ODE solution. (`plt.step(times, foxes, where="post")`
draws a population that jumps, rather than one that slopes.) Run it a few more times
with different seeds. How does a KMC trajectory differ from the ODE solution?

````python
def kmc(R0, F0, t_end, rng):
    """Rejection-free kinetic Monte Carlo for rabbits and foxes.
    Returns times, rabbits, foxes: NumPy arrays with one entry per event."""
    t, R, F = 0.0, R0, F0
    times, rabbits, foxes = [t], [R], [F]

    while t < t_end:
        # Step 2: the four rates in the current state
        r1 = k1 * R          # rabbit born
        r2 = k2 * R * F      # rabbit eaten
        r3 = k3 * R * F      # fox born
        r4 = k4 * F          # fox dies
        Q = r1 + r2 + r3 + r4

        # Step 3: if nothing can happen, stop
        if Q == 0:
            break

        # Step 4: pick an event. 1 - rng.random() is in (0, 1]
        uQ = (1 - rng.random()) * Q
        if uQ <= r1:
            R += 1
        elif uQ <= r1 + r2:
            R -= 1
        elif uQ <= r1 + r2 + r3:
            F += 1
        elif uQ <= Q:
            F -= 1
        else:
            raise RuntimeError(f"No event chosen: uQ={uQ}, Q={Q}")

        # Let bugs shout
        if R < 0 or F < 0:
            raise RuntimeError(f"Negative population at t={t}: R={R}, F={F}")

        # Step 6: advance the clock by an exponentially distributed time
        t += np.log(1 / (1 - rng.random())) / Q

        times.append(t)
        rabbits.append(R)
        foxes.append(F)

    return np.array(times), np.array(rabbits), np.array(foxes)


# One run, timed
rng = np.random.default_rng(1)
start = time.perf_counter()
times, rabbits, foxes = kmc(R0, F0, t_end, rng)
elapsed = time.perf_counter() - start
print(f"{len(times) - 1} events in {elapsed:.2f} s")

plt.plot(t_ode, R_ode, "k--", lw=1, label="rabbits (ODE)")
plt.plot(t_ode, F_ode, "k:", lw=1, label="foxes (ODE)")
plt.step(times, rabbits, where="post", label="rabbits (KMC)")
plt.step(times, foxes, where="post", label="foxes (KMC)")
plt.xlabel("time (days)")
plt.ylabel("population")
plt.title("Seed 1")
plt.legend()
plt.show()

# A few more seeds
for seed in [2, 3, 4]:
    times, rabbits, foxes = kmc(R0, F0, t_end, np.random.default_rng(seed))
    print(f"seed {seed}: {len(times) - 1} events, "
          f"{rabbits[-1]} rabbits and {foxes[-1]} foxes at the end")
    plt.plot(t_ode, F_ode, "k:", lw=1, label="foxes (ODE)")
    plt.step(times, rabbits, where="post", label="rabbits (KMC)")
    plt.step(times, foxes, where="post", label="foxes (KMC)")
    plt.xlabel("time (days)")
    plt.ylabel("population")
    plt.title(f"Seed {seed}")
    plt.legend()
    plt.show()
````

*Answer:*The stochastic simulation matches the ODE through the first cycle: across all seeds, foxes peaked around days 30–35 at roughly 2,500–3,000, close to the ODE's 2,804. The models diverge at the trough. The ODE lets foxes fall to 1.6 and recover, but in the simulation, whole foxes can't survive at such low numbers. In most runs (seeds 2, 3, 4, 6, 8), foxes went extinct between days 265 and 335. Without predators, rabbits then grew exponentially into the thousands by day 600. In the other runs, one or two foxes survived and started a second cycle.
