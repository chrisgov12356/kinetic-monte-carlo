Same farm, same rate constants: 400 rabbits and 200 foxes at the start, and

$$\frac{dR}{dt} = k_1 R - k_2 R F \qquad\qquad \frac{dF}{dt} = k_3 R F - k_4 F$$

with $k_1 = 0.015$ day$^{-1}$, $k_2 = 0.00004$ day$^{-1}$ foxes$^{-1}$,
$k_3 = 0.0004$ day$^{-1}$ rabbits$^{-1}$ and $k_4 = 0.04$ day$^{-1}$.

In the `solve_ivp` homework you found that the foxes fall to **under two** in each
trough. The ODEs don't mind: 1.6 foxes carry on and breed. Real foxes can't be
fractional, and two foxes can easily both die before they meet a rabbit. So now we
treat every rabbit and fox as an individual, and every birth and death as a random
event. That is **kinetic Monte Carlo** (KMC).

### The events

There are four events. Each has a rate, in events per day, which changes as the
populations change:

| Event | Rate | Effect |
|---|---|---|
| a rabbit is born | $k_1 R$ | $R \to R + 1$ |
| a rabbit is eaten | $k_2 R F$ | $R \to R - 1$ |
| a fox is born | $k_3 R F$ | $F \to F + 1$ |
| a fox dies | $k_4 F$ | $F \to F - 1$ |

The rates are easy to calculate, so we use the **rejection-free** algorithm, which you
worked out in class from
[Wikipedia's Kinetic Monte Carlo page](https://en.wikipedia.org/wiki/Kinetic_Monte_Carlo).
Writing it down in your own words is the first part of task 1.

````python
import time

import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

k1 = 0.015  # 1/day
k2 = 0.00004  # 1/day/fox
k3 = 0.0004  # 1/day/rabbit
k4 = 0.04  # 1/day

R0, F0 = 400, 200  # rabbits and foxes at t = 0
t_end = 600  # days
````

The deterministic answer, for comparison. This is the `solve_ivp` homework's
solution, at a tight tolerance.

````python
def dydt(t, y):
    R, F = y
    return [k1 * R - k2 * R * F, k3 * R * F - k4 * F]

ode = solve_ivp(dydt, (0, t_end), [R0, F0], rtol=1e-10, atol=1e-8, dense_output=True)
t_ode = np.linspace(0, t_end, 3001)
R_ode, F_ode = ode.sol(t_ode)

plt.plot(t_ode, R_ode, label="rabbits (ODE)")
plt.plot(t_ode, F_ode, label="foxes (ODE)")
plt.xlabel("time (days)")
plt.ylabel("population")
plt.legend()
plt.show()
````

## 1. One trajectory

**(a) The algorithm, in your own words.** Before you write any code, write the
rejection-free algorithm as numbered steps, for rabbits and foxes: what each of
Wikipedia's symbols ($k$, $r_{ki}$, $R_{ki}$, $Q_k$, $u$, $u'$) becomes here, how the
next event is chosen, and how far the clock moves. Then work the **first step by
hand**, from 400 rabbits and 200 foxes, with $u = 0.5$ and $u' = 0.3$: which event
happens, and how many minutes pass?

*Answer:*

````python
1. k is the current state of the system, i.e the numbers of rabbits and foxes right now, (R,F)
2.r_ki is the rate of event i in that state. Since we have four events, i = 1 to 4. The events are: rabbits reproduce, foxes eat rabbits, foxes reproduce, and foxes die. The rates of these events are given by the equations:Rabbit reproduce: r_k1 = k1 * R, rabbits eaten by foxes: r_k2 = k2 * R * F, foxes reproduce: r_k3 = k3 * R * F, foxes die: r_k4 = k4 * F. The total rate of all events is the sum of the individual rates: r_total = r_k1 + r_k2 + r_k3 + r_k4
3.R_ki is the cumulative rate: R_k1 = r_k1, R_k2 = r_k1 + r_k2, and so on.
4.Q_k is the total rate of all events, which is the last cumulative rate, R_k4.
5. is a random number in (0, 1] that picks which event happens.
6. u′ is a second random number in (0, 1] that picks how long until it happens.
The steps:
1.t=0
k = (R0, F0) = (400, 200)  # Initial state with R0 rabbits and F0 foxes
2.Calculate the four rates and their cumulative sums
3. Get Q, the total rate. If Q = 0, stop.
4. Draw u. Pick the event i where R_k,i−1 < uQ ≤ R_ki. Imagine laying the four rates end to end on a ruler of length Q and seeing where uQ lands. Faster events take up more of the ruler, so they get picked more often.
5.Apply that event: add or remove one rabbit or fox.
6.Draw u′ and move the clock forward by Δt = ln(1/u′)/Q. A higher total rate means shorter waits.
7. If t < t_end, go back to step 2.

At R=400 and F=200, k1 = 0.015  # 1/day
k2 = 0.00004  # 1/day/fox
k3 = 0.0004  # 1/day/rabbit
k4 = 0.04  # 1/day the rates are:
- Rabbit reproduce: r_k1 = k1 * R = 0.015 * 400 = 6.0
- Rabbits eaten by foxes: r_k2 = k2 * R * F = 0.00004 * 400 * 200 = 3.2
- Foxes reproduce: r_k3 = k3 * R * F = 0.0004 * 400 * 200 = 32.0
- Foxes die: r_k4 = k4 * F = 0.04 * 200 = 8.0     
cumulative rates, R_ki = 
- R_k1 = r_k1 = 6.0
- R_k2 = r_k1 + r_k2 = 6.0 + 3.2 = 9.2
- R_k3 = r_k1 + r_k2 + r_k3 = 9.2 + 32.0 = 41.2
- R_k4 = r_k1 + r_k2 + r_k3 + r_k4 = 41.2 + 8.0 = 49.2
Q_k = R_k4 = 49.2 events/day
uQ = 0.5 * 49.2 = 24.6, That falls between 9.2 and 41.2, so a fox is born, and F goes from 200 to 201.
Δt = ln(1/0.3)/49.2 = 0.0203 days 
0.0203 days is about 29.2 minutes, so the new state is (R, F) = (400, 201) at t = 0.0203 days.
````

**(b) The function.** Write

```python
def kmc(R0, F0, t_end, rng):
    ...
    return times, rabbits, foxes   # three NumPy arrays, one entry per event
```

that runs your algorithm from $t = 0$ until the clock passes `t_end`. If the total
rate is ever zero, nothing more can happen, so stop there.

Run it once. How many events were there, and how long did it take? Plot rabbits and
foxes against time, on top of the ODE solution. (`plt.step(times, foxes, where="post")`
draws a population that jumps, rather than one that slopes.) Run it a few more times
with different seeds. How does a KMC trajectory differ from the ODE solution?

````python
def kmc(R0, F0, t_end, rng):
    """Rejection-free kinetic Monte Carlo for rabbits and foxes.
    Returns times, rabbits, foxes: NumPy arrays with one entry per event."""
    t, R, F = 0.0, R0, F0
    times, rabbits, foxes = [t], [R], [F]

    while t < t_end:
        # Step 2: the four rates in the current state
        r1 = k1 * R          # rabbit born
        r2 = k2 * R * F      # rabbit eaten
        r3 = k3 * R * F      # fox born
        r4 = k4 * F          # fox dies
        Q = r1 + r2 + r3 + r4

        # Step 3: if nothing can happen, stop
        if Q == 0:
            break

        # Step 4: pick an event. 1 - rng.random() is in (0, 1]
        uQ = (1 - rng.random()) * Q
        if uQ <= r1:
            R += 1
        elif uQ <= r1 + r2:
            R -= 1
        elif uQ <= r1 + r2 + r3:
            F += 1
        elif uQ <= Q:
            F -= 1
        else:
            raise RuntimeError(f"No event chosen: uQ={uQ}, Q={Q}")

        # Let bugs shout
        if R < 0 or F < 0:
            raise RuntimeError(f"Negative population at t={t}: R={R}, F={F}")

        # Step 6: advance the clock by an exponentially distributed time
        t += np.log(1 / (1 - rng.random())) / Q

        times.append(t)
        rabbits.append(R)
        foxes.append(F)

    return np.array(times), np.array(rabbits), np.array(foxes)


# One run, timed
rng = np.random.default_rng(1)
start = time.perf_counter()
times, rabbits, foxes = kmc(R0, F0, t_end, rng)
elapsed = time.perf_counter() - start
print(f"{len(times) - 1} events in {elapsed:.2f} s")

plt.plot(t_ode, R_ode, "k--", lw=1, label="rabbits (ODE)")
plt.plot(t_ode, F_ode, "k:", lw=1, label="foxes (ODE)")
plt.step(times, rabbits, where="post", label="rabbits (KMC)")
plt.step(times, foxes, where="post", label="foxes (KMC)")
plt.xlabel("time (days)")
plt.ylabel("population")
plt.title("Seed 1")
plt.legend()
plt.show()

# A few more seeds
for seed in [2, 3, 4]:
    times, rabbits, foxes = kmc(R0, F0, t_end, np.random.default_rng(seed))
    print(f"seed {seed}: {len(times) - 1} events, "
          f"{rabbits[-1]} rabbits and {foxes[-1]} foxes at the end")
    plt.plot(t_ode, F_ode, "k:", lw=1, label="foxes (ODE)")
    plt.step(times, rabbits, where="post", label="rabbits (KMC)")
    plt.step(times, foxes, where="post", label="foxes (KMC)")
    plt.xlabel("time (days)")
    plt.ylabel("population")
    plt.title(f"Seed {seed}")
    plt.legend()
    plt.show()
````

*Answer:*The stochastic simulation matches the ODE through the first cycle: across all seeds, foxes peaked around days 30–35 at roughly 2,500–3,000, close to the ODE's 2,804. The models diverge at the trough. The ODE lets foxes fall to 1.6 and recover, but in the simulation, whole foxes can't survive at such low numbers. In most runs (seeds 2, 3, 4, 6, 8), foxes went extinct between days 265 and 335. Without predators, rabbits then grew exponentially into the thousands by day 600. In the other runs, one or two foxes survived and started a second cycle.

## 3. The second fox peak

The ODE's foxes peak at about day 32, and again at about day 428, at about 2804
foxes each time. When is the second peak in a KMC trajectory, and how high?

A noisy trajectory has lots of tiny local maxima, so we need a definition. Use this
one: **the largest number of foxes after day 200, counted as a peak only if it is
above 100 foxes and isn't within 5 days of either end of the window** (day 200 or
day 600). Write a function `second_peak(times, foxes)` that returns `(time, height)`,
or `None` if there isn't one.

Run 100 trajectories. In what fraction is there a second peak? For those that have
one, give the mean and median of its time and of its height, and the interquartile
range (25th to 75th percentile, `np.percentile`) of each. Plot histograms.

How do these compare with the ODE's second peak? If they differ, why? And what does
the definition get wrong? Find a trajectory where it picks something you wouldn't
call "the second peak", if you can.

````python
def second_peak(times, foxes, t_start=200, t_stop=600, min_height=100, margin=5):
    """Largest number of foxes between t_start and t_stop.
    Returns (time, height), or None if it is not above min_height
    or is within `margin` days of either end of the window."""
    window = (times >= t_start) & (times <= t_stop)
    if not np.any(window):
        return None
    t_window, f_window = times[window], foxes[window]
    i = np.argmax(f_window)
    t_peak, height = t_window[i], f_window[i]
    if height <= min_height:
        return None
    if t_peak < t_start + margin or t_peak > t_stop - margin:
        return None
    return t_peak, height


n_runs = 100
peak_times, peak_heights = [], []
for seed in range(n_runs):
    times, rabbits, foxes = kmc(R0, F0, t_end, np.random.default_rng(seed))
    peak = second_peak(times, foxes)
    if peak is not None:
        peak_times.append(peak[0])
        peak_heights.append(peak[1])

peak_times = np.array(peak_times)
peak_heights = np.array(peak_heights)
n_peaks = len(peak_times)
print(f"Second peak in {n_peaks} of {n_runs} runs ({n_peaks / n_runs:.0%})")

for name, values, unit in [("time", peak_times, "days"), ("height", peak_heights, "foxes")]:
    q25, q75 = np.percentile(values, [25, 75])
    print(f"Peak {name}: mean {values.mean():.0f}, median {np.median(values):.0f}, "
          f"IQR {q25:.0f} to {q75:.0f} {unit}")

# The ODE's second peak, for comparison
mask = t_ode > 200
i = np.argmax(F_ode[mask])
print(f"ODE second peak: day {t_ode[mask][i]:.0f}, {F_ode[mask][i]:.0f} foxes")

fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].hist(peak_times, bins=15)
axes[0].axvline(t_ode[mask][i], color="k", ls="--", label="ODE")
axes[0].set_xlabel("second peak time (days)")
axes[0].set_ylabel("number of runs")
axes[0].legend()
axes[1].hist(peak_heights, bins=15)
axes[1].axvline(F_ode[mask][i], color="k", ls="--", label="ODE")
axes[1].set_xlabel("second peak height (foxes)")
axes[1].legend()
plt.tight_layout()
plt.show()

# A trajectory to look at closely for the last question
times, rabbits, foxes = kmc(R0, F0, t_end, np.random.default_rng(183))
peak = second_peak(times, foxes)
plt.step(times, foxes, where="post", label="foxes (KMC, seed 183)")
plt.plot(t_ode, F_ode, "k:", lw=1, label="foxes (ODE)")
plt.plot(*peak, "ro", label="picked as second peak")
plt.xlabel("time (days)")
plt.ylabel("foxes")
plt.legend()
plt.show()
````

*Answer:*Only runs whose foxes survive the trough can have a second peak, so the sample is biased. The survivors I inspected (seeds 11, 27, 56) bottomed out at 15–25 foxes, not the ODE’s 1.6. In this model, a cycle that swings less far from the balance point (100 rabbits, 375 foxes) has a shorter period and a lower peak. Survival therefore favors smaller cycles, which peak sooner and lower. Randomness also spreads out the timing, so no single run matches the ODE. The definition takes the largest value after day 200, which isn’t necessarily the second peak. If a third cycle rose higher before day 600, it would be picked instead. Seed 183 shows this risk: its second peak is at day 339, and its foxes are climbing again by day 600. A real peak that is still rising near day 595 gets discarded, even though it was only cut off by the end of the simulation. Noise also matters. The maximum lands on a random spike, which slightly inflates the height and makes the timing jitter. Finally, the cutoffs (day 200, 100 foxes, 5 days) are arbitrary, and changing them would change the results

## 4. How many runs are enough?

Every number in task 3 came from random runs, so it has its own uncertainty. Repeat
task 3 with 100, 300, 1000 and 3000 trajectories. (Time it first, and work out how
long 3000 will take before you start.)

- How do the mean second-peak time and height change with the number of runs?
- Work out the **standard error** of each mean, $s / \sqrt{n}$, where $s$ is the
  standard deviation and $n$ the number of trajectories with a peak. How does it
  scale with $n$?
- Give the expected time and height of the second peak, and their interquartile
  ranges, to a precision you can justify. Say how you justified it.

````python
# Time it first
start = time.perf_counter()
for seed in range(10):
    kmc(R0, F0, t_end, np.random.default_rng(seed))
per_run = (time.perf_counter() - start) / 10
print(f"{per_run:.3f} s per run, so 3000 runs should take about {per_run * 3000 / 60:.1f} minutes")

# Run 3000 trajectories once, storing the second peak and extinction time of each
N = 3000
all_peak_times = np.full(N, np.nan)
all_peak_heights = np.full(N, np.nan)
extinction = np.full(N, np.nan)   # used again in task 5

start = time.perf_counter()
for seed in range(N):
    times, rabbits, foxes = kmc(R0, F0, t_end, np.random.default_rng(seed))
    peak = second_peak(times, foxes)
    if peak is not None:
        all_peak_times[seed], all_peak_heights[seed] = peak
    if np.any(foxes == 0):
        extinction[seed] = times[np.argmax(foxes == 0)]
print(f"3000 runs took {(time.perf_counter() - start) / 60:.1f} minutes")

# Statistics for the first 100, 300, 1000 and 3000 runs
run_counts = [100, 300, 1000, 3000]
se_times, se_heights, n_with_peak = [], [], []
for n_runs in run_counts:
    t = all_peak_times[:n_runs]
    h = all_peak_heights[:n_runs]
    t, h = t[~np.isnan(t)], h[~np.isnan(h)]
    n = len(t)
    se_t = t.std(ddof=1) / np.sqrt(n)
    se_h = h.std(ddof=1) / np.sqrt(n)
    se_times.append(se_t)
    se_heights.append(se_h)
    n_with_peak.append(n)
    t25, t75 = np.percentile(t, [25, 75])
    h25, h75 = np.percentile(h, [25, 75])
    print(f"{n_runs:5d} runs, {n:4d} with a peak | "
          f"time {t.mean():.1f} ± {se_t:.1f} days, IQR {t25:.0f} to {t75:.0f} | "
          f"height {h.mean():.0f} ± {se_h:.0f} foxes, IQR {h25:.0f} to {h75:.0f}")

# How does the standard error scale?
n_with_peak = np.array(n_with_peak)
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
for ax, se, label in [(axes[0], se_times, "time (days)"), (axes[1], se_heights, "height (foxes)")]:
    ax.loglog(n_with_peak, se, "o-", label="standard error")
    ax.loglog(n_with_peak, se[0] * np.sqrt(n_with_peak[0] / n_with_peak), "k--",
              label=r"$\propto 1/\sqrt{n}$")
    ax.set_xlabel("n (runs with a second peak)")
    ax.set_ylabel(f"standard error of mean {label}")
    ax.legend()
plt.tight_layout()
plt.show()
````

*Answer:*The means wander at first and then settle. The peak time stays within a few days of 400 throughout, while the height drops from about 2,500 at 100 runs to about 2,390 at 3,000. The 100-run mean, based on only 33 peaks, was further off than its spread alone would suggest. Every change between rows falls within about two standard errors, which is consistent with noise.
Each tripling of the number of runs cuts the standard error by about √3 ≈ 1.7 (5.1 → 2.8 → 1.7 → 0.9 days). This is 1/√n scaling, and on the log-log plot the points follow the dashed line. In practice, halving the uncertainty takes four times as many runs.
With 3,000 runs, the second peak occurs at about 402 ± 2 days, with a height of about 2,390 ± 40 foxes. Both uncertainties are two standard errors, roughly a 95% confidence interval. The interquartile ranges are about 380–420 days and 2,000–2,700 foxes. They deserve less precision than the means, because their edges shifted by about 5 days and 50–100 foxes between the 1,000- and 3,000-run sets.
